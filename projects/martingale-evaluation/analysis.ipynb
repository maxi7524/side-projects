```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

def simulate_martingales(n_steps, n_paths):
    """
    Simulates random walk paths and the associated martingale Y_n = X_n^2 - n.
    """
    # Generate random steps: 1 for heads, -1 for tails
    steps = np.random.choice([-1, 1], size=(n_paths, n_steps))
    
    # X_n: Cumulative sum of steps (Wealth of the player)
    # Adding a starting column of zeros for time t=0
    x_n = np.hstack([np.zeros((n_paths, 1)), np.cumsum(steps, axis=1)])
    
    # Y_n = X_n^2 - n
    time_indices = np.arange(n_steps + 1)
    y_n = x_n**2 - time_indices
    
    return time_indices, x_n, y_n

def plot_results(time, x_paths, y_paths):
    """
    Plots the trajectories of both processes and their average values.
    """
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10), sharex=True)
    
    # Plot X_n (Simple Random Walk)
    ax1.plot(time, x_paths.T, color='blue', alpha=0.1)
    ax1.plot(time, np.mean(x_paths, axis=0), color='red', linewidth=2, label='Mean of $X_n$ (Expected: 0)')
    ax1.set_title("Simple Random Walk $X_n$ (Martingale)")
    ax1.set_ylabel("Wealth ($)")
    ax1.legend()
    ax1.grid(True)

    # Plot Y_n = X_n^2 - n
    ax2.plot(time, y_paths.T, color='green', alpha=0.1)
    ax2.plot(time, np.mean(y_paths, axis=0), color='red', linewidth=2, label='Mean of $Y_n$ (Expected: 0)')
    ax2.set_title("Martingale $Y_n = X_n^2 - n$")
    ax2.set_xlabel("Step (n)")
    ax2.set_ylabel("Value")
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

def simulate_conditional_martingale(n_steps=2000, n_paths=5000, check_step=1000):
    """
    Simulates paths and focuses on the conditional expectation property.
    """
    # 1. Generate steps (-1, 1)
    steps = np.random.choice([-1, 1], size=(n_paths, n_steps))
    x_n = np.cumsum(steps, axis=1)
    
    # 2. Find paths that are at a specific 'profit' at check_step
    # Let's pick paths that are significantly above zero to see if they 'fall' or stay
    target_value = 30 
    tolerance = 5
    mask = (x_n[:, check_step] >= target_value - tolerance) & \
           (x_n[:, check_step] <= target_value + tolerance)
    
    filtered_paths = x_n[mask]
    
    return np.arange(n_steps), filtered_paths, check_step, target_value

def plot_conditional_behavior(time, paths, check_step, target_val):
    plt.figure(figsize=(12, 6))
    
    if len(paths) == 0:
        print("No paths found in that range, try lower target or more paths.")
        return

    # Plot the filtered paths from the check_step onwards
    for i in range(min(len(paths), 100)): # Plot up to 100 paths for clarity
        plt.plot(time[check_step:], paths[i, check_step:], color='blue', alpha=0.05)
    
    # Calculate the mean of these specific paths
    path_mean = np.mean(paths, axis=0)
    
    plt.axvline(x=check_step, color='black', linestyle='--', label=f'Point of observation (n={check_step})')
    plt.plot(time[check_step:], path_mean[check_step:], color='red', linewidth=3, 
             label='Conditional Mean (Expectation)')
    
    plt.axhline(y=path_mean[check_step], color='green', linestyle=':', label=f'Value at n={check_step}')
    
    plt.title(f"Martingale Property: E[X_m | X_n] = X_n")
    plt.xlabel("Steps")
    plt.ylabel("Wealth")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
    
```

```python
N_STEPS = 5000   # Number of coin flips
N_PATHS = 20   # Number of independent simulations

time, x_data, y_data = simulate_martingales(N_STEPS, N_PATHS)
plot_results(time, x_data, y_data)
```

```python
# Execution
time, filtered_paths, check_step, target_val = simulate_conditional_martingale(N_STEPS, N_PATHS, 1000)
plot_conditional_behavior(time, filtered_paths, check_step, target_val)
```
